## Exercice 1.1 : la transposée

Écrivez `transposee(A)` qui échange lignes et colonnes.

**Résultat attendu :** la transposée de `[[1, 2, 3], [4, 5, 6]]` est
`[[1, 4], [2, 5], [3, 6]]`.


```python
# votre réponse ici
def transposee(A):
    lignes = len(A)
    colonnes = len(A[0])
  
    T = [[A[l][c] for l in range(lignes)] for c in range(colonnes)]
    return T

print(transposee([[1, 2, 3], [4, 5, 6]]))
```

## Exercice 2.1 : produit à la main

Écrivez `produit(A, B)`, et vérifiez sur un exemple que vous savez calculer de tête.

**Résultat attendu :** multiplier une matrice par l'identité la laisse inchangée.


```python
# votre réponse ici

def produit(A, B):
    l_A = len(A)
    c_A = len(A[0])
    l_B = len(B)
    c_B = len(B[0])
    
  
    res = [[0 * c_B for _ in range(c_B)] for _ in range(l_A)]
    
    for i in range(l_A):
        for j in range(c_B):
            for k in range(c_A):
                res[i][j] += A[i][k] * B[k][j]
    return res


```

## Exercice 6.1 : composer deux transformations

Vérifiez numériquement que faire tourner de 30 degrés puis de 60 degrés revient à
faire tourner de 90 degrés, autrement dit que
$R(60°) \cdot R(30°) = R(90°)$.


```python
# votre réponse ici

import math

def rotation(deg):
    rad = math.radians(deg)
    c, s = math.cos(rad), math.sin(rad)
    return [[c, -s, 0], [s, c, 0], [0, 0, 1]]

R30 = rotation(30)
R60 = rotation(60)
R90 = rotation(90)

prod = produit(R60, R30)
print("R(60) * R(30) égal à R(90) ? (comparaison approximative)")
print("Matrice produit R60*R30 :", [[round(x, 3) for x in ligne] for ligne in prod])
```

## Exercice 7.1 : déterminant par Gauss

Écrivez `determinant(A)` en reprenant l'élimination de `resoudre`, sans second membre :
multipliez les pivots successifs, et changez le signe à chaque échange de lignes.
Comparez à `np.linalg.det`.

**Résultat attendu :** le déterminant de la matrice de l'étape 1 vaut `-2`, et celui
de `[[1, 2], [2, 4]]` vaut `0`.


```python
# votre réponse ici
def determinant(A):
    n = len(A)
    M = [row[:] for row in A]
    det = 1.0
    signe = 1
    
    for i in range(n):
       
        max_el = abs(M[i][i])
        max_row = i
        for k in range(i + 1, n):
            if abs(M[k][i]) > max_el:
                max_el = abs(M[k][i])
                max_row = k
                
        if M[max_row][i] == 0:
            return 0.0
            
        if max_row != i:
            M[i], M[max_row] = M[max_row], M[i]
            signe *= -1
            
        det *= M[i][i]
        
        for k in range(i + 1, n):
            c = M[k][i] / M[i][i]
            for j in range(i, n):
                M[k][j] -= c * M[i][j]
                
    return signe * det

print("Déterminant [[1, 2], [2, 4]] :", determinant([[1, 2], [2, 4]]))
```
